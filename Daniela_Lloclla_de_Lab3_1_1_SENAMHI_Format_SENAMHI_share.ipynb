### Ejercicio 2: Datos climáticos multiestación

Crea un nuevo DataFrame con datos de 5 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Luego:
1. Paresear el nombres de la estaciones y leer los archivo.
1. Unir las db (Fecha, Est1, Est2, ...)
1. Calcula el promedio mensual de precipitación por región
2. Identifica el mes más lluvioso en cada estación
4. Exporta el resultado formateado a CSV

```python

# Leer los datos de las estaciones
import os
import pandas as pd

files = os.listdir('/content/drive/MyDrive/Senamhi')
files

# Parsear los nombres de las estaciones y leer los archivos

estaciones = []

for station in files:

    # Solo leer los archivos .txt
    if station.endswith('.txt'):

        # Parsear el nombre del archivo
        codigo = station.split('_')[0]
        nombre = station.split('_')[1].replace('.txt', '')

        print(f"Código: {codigo}, Nombre: {nombre}")

        # Leer el archivo
        station_df = pd.read_table(
            f'/content/drive/MyDrive/Senamhi/{station}',
            header=None,
            sep=' ',
            na_values=['-999', '-99.9', '-3256']
        )

        # Asignar nombres de columnas
        station_df.columns = ['Año', 'Mes', 'Dia', 'Pr', 'Tmax', 'Tmin']

        # Formatear las columnas
        station_df['Año'] = station_df['Año'].astype(int)
        station_df['Mes'] = station_df['Mes'].astype(int)
        station_df['Dia'] = station_df['Dia'].astype(int)

        # Crear la fecha
        station_df['Fecha'] = pd.to_datetime(
            station_df[['Año', 'Mes', 'Dia']].rename(
                columns={
                    'Año': 'year',
                    'Mes': 'month',
                    'Dia': 'day'
                }
            )
        )

        # Guardar la estación
        estaciones.append(station_df)

# Revisar las estaciones

len(estaciones)

# Unir las bases de datos
# Fecha, estación 1, estación 2, etc.

clima_df = pd.DataFrame()

clima_df['Fecha'] = estaciones[0]['Fecha']
clima_df['Ñaña'] = estaciones[0]['Pr']
clima_df['Chosica'] = estaciones[1]['Pr']
clima_df['Santiago_de_Tuna'] = estaciones[2]['Pr']
clima_df['Canchacalla'] = estaciones[3]['Pr']
clima_df['Santa_Eulalia'] = estaciones[4]['Pr']

clima_df

# Promedio mensual de precipitación por estación

precipitacion_mensual = clima_df.groupby(
    clima_df['Fecha'].dt.month
)[
    ['Ñaña', 'Chosica', 'Santiago_de_Tuna', 'Canchacalla', 'Santa_Eulalia']
].mean()

print(f"Promedio mensual de precipitación:\n{precipitacion_mensual}")

# Identificar el mes más lluvioso en cada estación

mes_mas_lluvioso = precipitacion_mensual.idxmax()

print(f"Mes más lluvioso en cada estación:\n{mes_mas_lluvioso}")

# Mostrar el número del mes más lluvioso y su precipitación promedio

for estacion in precipitacion_mensual.columns:

    mes = precipitacion_mensual[estacion].idxmax()
    precipitacion = precipitacion_mensual[estacion].max()

    print(
        f"{estacion}: mes {mes}, "
        f"precipitación promedio {precipitacion:.2f} mm"
    )

# Exportar el promedio mensual a CSV

precipitacion_mensual.to_csv(
    '/content/drive/MyDrive/Senamhi/precipitacion_mensual.csv'
)

# Exportar el mes más lluvioso a CSV

mes_mas_lluvioso.to_csv(
    '/content/drive/MyDrive/Senamhi/mes_mas_lluvioso.csv'
)

```
